# 02 - Sistemas operacionais

Consultas sobre `survey_operating_systems` no DuckDB.

```python
import duckdb
from pathlib import Path

db = Path('..') / 'data' / 'processed' / 'project.duckdb'
con = duckdb.connect(str(db), read_only=True)
con.execute('''
    SELECT survey_year, operating_system_family, count(DISTINCT respondent_id) AS respondents
    FROM survey_operating_systems
    GROUP BY survey_year, operating_system_family
    ORDER BY survey_year, respondents DESC
''').df()
```
