# 02 - Sistemas operacionais

Consultas sobre `survey_operating_systems` no DuckDB.

```python
import duckdb
from IPython.display import display
from pathlib import Path

def find_project_root(start=Path.cwd()):
    for path in (start, *start.parents):
        if (path / 'src').exists() and (path / 'data').exists():
            return path
    raise RuntimeError('Abra o notebook dentro do projeto os-data-project.')

root = find_project_root()
db = root / 'data' / 'processed' / 'project.duckdb'
if not db.exists():
    raise FileNotFoundError(f'Banco DuckDB nao encontrado: {db}. Rode src/build_database.py antes.')

print(f'Banco usado: {db}')
con = duckdb.connect(str(db), read_only=True)
df = con.execute('''
    SELECT survey_year, operating_system_family, count(DISTINCT respondent_id) AS respondents
    FROM survey_operating_systems
    GROUP BY survey_year, operating_system_family
    ORDER BY survey_year, respondents DESC
''').df()
con.close()

print(f'Linhas retornadas: {len(df)}')
print(df.to_string(index=False))

display(df)
```
