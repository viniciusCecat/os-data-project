```python
from pathlib import Path
import csv

def find_project_root(start=Path.cwd()):
    for path in (start, *start.parents):
        if (path / 'src').exists() and (path / 'data').exists():
            return path
    raise RuntimeError('Abra o notebook dentro do projeto os-data-project.')

root = find_project_root()
for year in (2022, 2023, 2024, 2025):
    path = root / 'data' / 'raw' / 'stackoverflow' / f'year={year}' / 'results.csv'
    print(year, path.exists(), path)
    if path.exists():
        with path.open(encoding='utf-8-sig', newline='', errors='replace') as f:
            print(next(csv.reader(f))[:20])
```
